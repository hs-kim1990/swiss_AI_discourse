```python
# change center-right to right-wing, center-left to left-wing
df_media['political_orientation'] = df_media['political_orientation'].replace({'center-right': 'right-wing', 'center-left': 'left-wing'})
```

```python
df_media.columns
```

```python
# map medium_name in df to media in df_media to get political_orientation
df = df.merge(df_media[[' media', 'political_orientation']], left_on='medium_name', right_on=' media', how='left')
df[['medium_name', 'political_orientation']].head()
```

```python
# map sentiment_score to sentiment_label 



import numpy as np

vals = df['sentiment_score'].values
cut1 = np.percentile(vals, 33)
cut2 = np.percentile(vals, 66)

def relative_color(score):
    if score <= cut1:
        return -1
    elif score >= cut2:
        return 1
    else:
        return 0
df['sentiment_label'] = df['sentiment_score'].apply(relative_color)
```

```python
# draw heatmap of political_orientation x topic labels with value of average of sentiment scores

import seaborn as sns
import matplotlib.pyplot as plt
heatmap_data = df.pivot_table(index='topic_name', columns='political_orientation', values='sentiment_label', aggfunc='mean')
# x label order (['left-wing', 'center-left', 'center', 'center-right', 'right-wing']) with bold y labels
heatmap_data = heatmap_data[['left-wing', 'center', 'right-wing']]
plt.figure(figsize=(12, 4))
# change cmap to green - red
sns.heatmap(heatmap_data, annot=True, cmap='RdYlGn', fmt='.2f')
# set y label to bold
plt.title('Average Sentiment Scores by Political Orientation and Topic Label')
plt.show()
```

```python
# line graph with time on x-axis, count on y-axis, separate lines for each political orientation
df['pub_date'] = pd.to_datetime(df['pubtime'].astype(str).str[:10], errors='coerce')
df['year_month'] = df['pub_date'].dt.to_period('M').dt.to_timestamp()
df_count = df.groupby(['year_month', 'political_orientation']).size().reset_index(name='count')
plt.figure(figsize=(12, 6))
sns.lineplot(data=df_count, x='year_month', y='count', hue='political_orientation')
plt.title('Number of Articles Over Time by Political Orientation')
plt.xlabel('Year-Month')
plt.ylabel('Number of Articles')
plt.show()
```

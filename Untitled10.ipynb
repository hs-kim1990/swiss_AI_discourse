```python
import re
import nltk
from nltk.corpus import stopwords
from collections import Counter

nltk.download("stopwords")

# 1. Base stopwords (German + English)
base_stop = set(stopwords.words("german")) | set(stopwords.words("english"))

# 2. Extra stopwords from your frequency list
extra_stopwords = {
    "prozent", "uhr", "mehr", "jahr", "jahre", "jahren",
    "gibt", "sei", "sagt", "sagte", "sagen",
    "immer", "schon", "seit", "neue", "neuen", "bereits",
    "viele", "viel", "heute", "gestern", "morgen",
    "beim", "etwa", "geht", "gut", "ja", "ab", "rund",
    "november", "oktober", "montag", "dienstag",
    "zeit", "ende",
    "franken", "millionen", "milliarden",
    "unternehmen", "menschen", "bild", "mitteilung",
    "müssen", "wurde", "wurden",
    "schweizer", "prozent", "uhr", "jahr", "jahre", "jahren",
    "gibt", "sei", "sagt", "sagte", "sagen",
    "immer", "schon", "seit", "neue", "neuen", "bereits",
    "viele", "viel", "heute", "gestern", "morgen",
    "beim", "etwa", "geht", "gut", "ja", "ab", "rund",
    "november", "oktober", "montag", "dienstag",
    "zeit", "ende",
    "franken", "millionen", "milliarden",
    "unternehmen", "menschen", "bild", "mitteilung",
    "müssen", "wurde", "wurden", "schweiz", "schweizer",
    "laut", "angaben", "dpa", "sda",
    "zudem", "damit", "dabei", "darüber",
    "erste", "ersten", "zweite", "zweiten",
    "soll", "sollen",
    "machen", "macht",
    "zwar", "jedoch", "dennoch", "weiter",
}

# 3. Important short tokens we want to keep
keep_short = {"ki", "ai", "llm", "eu", "usa", "ch"}

# 4. Final stopword set
stop_words = (base_stop | extra_stopwords) - keep_short

def clean_text(text):
    text = str(text).lower()
    # keep only letters and spaces
    text = re.sub(r"[^a-zA-ZäöüÄÖÜß ]", " ", text)
    tokens = text.split()
    # remove stopwords and very short words, but keep important short ones
    tokens = [
        t for t in tokens
        if t not in stop_words and (len(t) > 2 or t in keep_short)
    ]
    return " ".join(tokens)

# 5. Rebuild text_raw and text_clean
df["text_raw"] = df["head"].fillna("").astype(str) + " " + df["content_parsed"].fillna("").astype(str)
df["text_clean"] = df["text_raw"].apply(clean_text)

# 6. Inspect the new top words
Counter(" ".join(df["text_clean"]).split()).most_common(50)
```

```python
import pandas as pd
from sentence_transformers import SentenceTransformer
from sklearn.cluster import KMeans
from sklearn.feature_extraction.text import TfidfVectorizer

# 1. Use the cleaned text
texts = df["text_clean"].astype(str).tolist()
print("Number of texts:", len(texts))

# 2. Load multilingual BERT sentence transformer
model = SentenceTransformer("sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2")

# 3. Create embeddings
embeddings = model.encode(texts, show_progress_bar=True)

# 4. Cluster embeddings into topics
n_topics = 10  # change this number if you want more or fewer topics

kmeans = KMeans(n_clusters=n_topics, random_state=42, n_init=10)
topic_ids = kmeans.fit_predict(embeddings)

df["topic_id"] = topic_ids

# 5. Find top keywords for each topic using TF-IDF
topic_keywords = {}

for topic in range(n_topics):
    topic_texts = df.loc[df["topic_id"] == topic, "text_clean"]
    if topic_texts.empty:
        continue

    # TF-IDF on the texts of this topic
    vectorizer = TfidfVectorizer(max_df=0.95, min_df=2)
    X = vectorizer.fit_transform(topic_texts)

    # Average TF-IDF score per term
    scores = X.mean(axis=0).A1
    terms = vectorizer.get_feature_names_out()

    top_n = 10
    top_indices = scores.argsort()[::-1][:top_n]
    keywords = [terms[i] for i in top_indices]

    topic_keywords[topic] = ", ".join(keywords)

# 6. Attach human readable topic label
df["topic_label"] = df["topic_id"].map(topic_keywords)

# 7. Show topics and some examples
for topic, kw in topic_keywords.items():
    print(f"\nTopic {topic}: {kw}")
    sample = df[df["topic_id"] == topic][["head", "topic_label"]].head(3)
    print(sample.to_string(index=False))
```

```python
!pip install umap-learn

import umap
import numpy as np
import matplotlib.pyplot as plt

# 1. Use your BERT embeddings again
embeddings = model.encode(texts, show_progress_bar=True)

# 2. Reduce to 2D
reducer = umap.UMAP(n_neighbors=30, min_dist=0.0, metric='cosine')
embedding_2d = reducer.fit_transform(embeddings)

df["x"] = embedding_2d[:,0]
df["y"] = embedding_2d[:,1]

# 3. Aggregate bubble sizes
topic_sizes = df.groupby("topic_name").size()

plt.figure(figsize=(12,8))

for topic, size in topic_sizes.items():
    sub = df[df["topic_name"] == topic]
    plt.scatter(
        sub["x"].mean(),
        sub["y"].mean(),
        s=size*0.5, # scale bubbles
        label=topic,
        alpha=0.6
    )

plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.title("BERT Topic Bubble Chart (UMAP)")
plt.show()
```

```python
import torch
import numpy as np
from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch.nn.functional as F

model_name = "cardiffnlp/twitter-xlm-roberta-base-sentiment"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSequenceClassification.from_pretrained(model_name).to("cuda")

texts = df["text_clean"].fillna("").tolist()

scores = []
batch_size = 32

for i in range(0, len(texts), batch_size):
    batch = texts[i:i+batch_size]

    enc = tokenizer(
        batch,
        truncation=True,
        max_length=128,
        padding=True,
        return_tensors="pt"
    ).to("cuda")

    with torch.no_grad():
        logits = model(**enc).logits
        probs = F.softmax(logits, dim=1).cpu().numpy()

    # cardiffnlp order: [negative, neutral, positive]
    neg = probs[:, 0]
    pos = probs[:, 2]

    # continuous score
    score = pos - neg

    scores.extend(score.tolist())

    if i % 3200 == 0:
        print(f"Processed {i} / {len(texts)}")

len(scores)
```

```python
df["sentiment_score"] = scores
df["sentiment_score"].describe()

```

```python
topic_mean_sent = df.groupby("topic_name")["sentiment_score"].mean().sort_values()
topic_mean_sent
```

```python
min_s = topic_mean_sent.min()
max_s = topic_mean_sent.max()

mid_s = (min_s + max_s) / 2   # center point
```

```python
import numpy as np

vals = topic_mean_sent.values
cut1 = np.percentile(vals, 33)
cut2 = np.percentile(vals, 66)

def relative_color(score):
    if score <= cut1:
        return "red"
    elif score >= cut2:
        return "green"
    else:
        return "grey"
```

```python
topic_colors = topic_mean_sent.apply(relative_color)
topic_colors
```

```python
# Create x/y coordinates from UMAP
df["x"] = embedding_2d[:, 0]
df["y"] = embedding_2d[:, 1]

# Compute topic-level centers and sizes
topic_coords = df.groupby("topic_name")[["x", "y"]].mean()
topic_sizes = df.groupby("topic_name").size()
```
